```python
df["duration_value"] = (
    df["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)
```

```python
df["duration_type"] = df["duration_type"].replace({
    "min": "Minutes",
    "Season": "Seasons",
    "Seasons": "Seasons"
})
```

```python
df["main_genre"] = (
    df["listed_in"]
    .str.split(",")
    .str[0]
    .str.strip()
)
```

```python
df["content_age"] = (
    df["year_added"] - df["release_year"]
)
```

```python
df[[
    "release_year",
    "year_added",
    "content_age"
]].head()
```

```python
def release_era(year):
    if year < 1980:
        return "Before 1980"
    elif year < 2000:
        return "1980-1999"
    elif year < 2010:
        return "2000-2009"
    elif year < 2020:
        return "2010-2019"
    else:
        return "2020+"
```

```python
df["release_era"] = df["release_year"].apply(release_era)
```

```python
df[[
    "release_year",
    "release_era"
]].head(20)
```

```python
df["has_director"] = np.where(
    df["director"] == "Unknown",
    0,
    1
)
```

```python
df["has_country"] = np.where(
    df["country"] == "Unknown",
    0,
    1
)
```
