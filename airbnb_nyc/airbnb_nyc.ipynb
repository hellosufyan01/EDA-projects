```python
def detect_outliers_iqr(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))

    return outliers
```

```python
price_outliers = detect_outliers_iqr(df, "price")
```

```python
minimum_nights_outliers = detect_outliers_iqr(df, "minimum_nights")
```

```python
def availability_category(days):
    if days == 0:
        return "Never Available"
    elif days < 100:
        return "Rarely Available"
    elif days < 250:
        return "Occasionally Available"
    else:
        return "Highly Available"
```

```python
df["Availability_Category"] = df["availability_365"].apply(
    availability_category
)
```
