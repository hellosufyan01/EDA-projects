```python

def detect_outliers_irq(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower) | (data[column] > upper)]

    print("column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("lower:", lower)
    print("upper:", upper)
    print("outliers:", len(outliers))
    return outliers
     
```

```python
age_outliers = detect_outliers_irq(df, "age")
```

```python
spent_outliers = detect_outliers_irq(df, "total_spent")
```

```python
visits_outliers = detect_outliers_irq(df, "total_visits")
```

handle outliers - capping 

```python
def cap_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    data[column] = (data[column].clip(lower, upper))
```

```python
cap_outliers(df, "age")
```

```python
cap_outliers(df, "total_spent")
```

```python
cap_outliers(df, "total_visits")
```
